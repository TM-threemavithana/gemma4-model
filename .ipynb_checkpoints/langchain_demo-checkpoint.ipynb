# 🚀 Gemma 4 + LangChain & LangGraph

This notebook demonstrates how to connect **LangChain** and **LangGraph** to your local Gemma 4 server.

### Prerequisites
1. **Start the Gemma 4 server** in your terminal:
   ```bash
   cd ~/gemma-server && ./run.sh
   ```
2. **Install Dependencies**:
   ```bash
   pip install langchain langchain-openai langgraph
   ```

````python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Point ChatOpenAI at your local Gemma 4 server
GEMMA_BASE_URL = "http://localhost:8000/v1"

llm = ChatOpenAI(
    model="gemma-4-e2b-it",
    base_url=GEMMA_BASE_URL,
    api_key="not-needed",
    temperature=0.7,
    max_tokens=512,
)

print(f"Connected to local Gemma server at {GEMMA_BASE_URL}")
````

## 1. Basic Chat
A simple one-shot question to the model.

````python
response = llm.invoke("Explain quantum computing in 3 sentences.")
print(response.content)
````

## 2. System Prompting
Defining a persona for the assistant.

````python
messages = [
    SystemMessage(content="You are a helpful pirate assistant. Always answer in pirate speak."),
    HumanMessage(content="What is machine learning?"),
]
response = llm.invoke(messages)
print(response.content)
````

## 3. Streaming
Streaming tokens as they are generated.

````python
print("Streaming response:")
for chunk in llm.stream("Write a short poem about AI."):
    print(chunk.content, end="", flush=True)
````

## 4. LCEL Chain
Using LangChain Expression Language to build a reusable pipeline.

````python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert {domain} teacher. Explain concepts clearly and concisely."),
    ("human", "{question}"),
])

chain = prompt | llm | StrOutputParser()

result = chain.invoke({"domain": "physics", "question": "What is entropy?"})
print(result)
````

## 5. LangGraph Agent (ReAct)
A stateful agent that can use tools. This requires `langgraph` to be installed.

````python
from langgraph.prebuilt import create_react_agent
from langchain_core.tools import tool
from datetime import datetime

@tool
def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression, {"__builtins__": {}}, {}))
    except Exception as e:
        return f"Error: {e}"

@tool
def get_current_time() -> str:
    """Get the current date and time."""
    return datetime.now().strftime("%Y-%m-%d %H:%M:%S")

# Create the agent with tools
agent = create_react_agent(llm, tools=[calculator, get_current_time])

# Run the agent
query = "What is 42 * 17 + 123?"
result = agent.invoke({"messages": [HumanMessage(content=query)]})

for msg in result["messages"]:
    print(f"[{msg.type.upper()}] {msg.content}")
````

## 6. Multi-turn Conversation
Simulating a conversation with memory/history.

````python
history = [SystemMessage(content="You are a helpful coding assistant.")]

def chat(query):
    print(f"👤 User: {query}")
    history.append(HumanMessage(content=query))
    response = llm.invoke(history)
    print(f"🤖 Gemma: {response.content}")
    history.append(AIMessage(content=response.content))
    print("-" * 30)

chat("What is a Python decorator?")
chat("Can you show me a simple example?")
````
